```
import os
import pandas as pd
from sqlalchemy import create_engine
from dotenv import load_dotenv

# Dynamically find the project root regardless of who runs it
NOTEBOOK_DIR = os.getcwd()
PROJECT_ROOT = os.path.abspath(os.path.join(NOTEBOOK_DIR, '..', '..'))

GOLD_DIR = os.path.join(PROJECT_ROOT, 'data', 'gold')
ENV_PATH = os.path.join(PROJECT_ROOT, '.env')

load_dotenv(ENV_PATH)
db_uri = os.getenv("DATABASE_URL")
engine = create_engine(db_uri, connect_args={"connect_timeout": 60})
```

```
# ==============================================================================
# Table   : dim_players
# Layer   : Gold -> PostgreSQL
# Purpose : Upload Star Schema to Production Database
# ==============================================================================

print("=" * 80)
print("Uploading Table : dim_players")
print("=" * 80)
```

```
# ------------------------------------------------------------------------------
# Step 1 : Read Gold Dataset
# ------------------------------------------------------------------------------
table_name = "dim_players"
folder_path = os.path.join(GOLD_DIR, table_name)
csv_files = [f for f in os.listdir(folder_path) if f.endswith('.csv')]

if csv_files:
    df = pd.read_csv(os.path.join(folder_path, csv_files[0]))
    rows_read = len(df)
    print(f"Rows Read from CSV : {rows_read}")
else:
    df = None
    rows_read = 0
    print(f"No CSV found for {table_name}")
```

```
# ------------------------------------------------------------------------------
# Step 2 : Upload to Postgres
# ------------------------------------------------------------------------------
if df is not None:
    df.to_sql(table_name, engine, if_exists='replace', index=False)
```

```
# ------------------------------------------------------------------------------
# Step 3 : Upload Summary
# ------------------------------------------------------------------------------
if df is not None:
    print("\n" + "=" * 80)
    print("UPLOAD SUMMARY")
    print("=" * 80)
    print(f"Table                : dim_players")
    print(f"Rows Uploaded        : {rows_read}")
    print("Status               : SUCCESS")
    print("=" * 80)
```
